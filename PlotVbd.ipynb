# Plot Breakdown Voltages

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
Top1 = np.loadtxt('Batch_1/cold_tests/Data/IV/splitted/top/quadrants/Vbds.txt')
Bot1 = np.loadtxt('Batch_1/cold_tests/Data/IV/splitted/bottom/quadrants/quadrantsVbds.txt')
Top2 = np.loadtxt('Batch_2/cold_tests/Data/IV/splitted/top/quadrants/Vbds.txt')
Bot2 = np.loadtxt('Batch_2/cold_tests/Data/IV/splitted/bottom/quadrants/Vbds.txt')
```

```python
vb92, vd75, vb94, vb65, vb85 = Bot1
vb51, vb48 = Top1
vb76, _, _, _ = Bot2
vb98, vb57, vb83 = Top2
```

```python
vbds = np.array((vb92, vd75, vb94, vb65, vb85, vb51, vb48, vb76,vb98, vb57, vb83))
```

```python
x = np.linspace(1, 4, 4)
vpdu = [92, 75, 94, 65, 85, 51, 48, 76, 98, 57, 83]
i = 0
plt.figure(figsize = [8,6])
plt.title('Breakdown Voltage', fontsize = 15)
for v in vbds:
    plt.plot(x, v, 'o', label = f'vPDU-{vpdu[i]}')
    i = i+1
plt.xticks(x)
plt.legend(loc="center left", bbox_to_anchor=(1, 0.5))
plt.xlabel('Quadrants', fontsize = 13)
plt.ylabel('Voltage (V)', fontsize = 13)
```
